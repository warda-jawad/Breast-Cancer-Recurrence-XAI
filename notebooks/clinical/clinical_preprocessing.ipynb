##Define Paths & Load Dataset

```python
# ============================================================
# Load Clinical Dataset
# ============================================================

DATA_DIR = "/content/drive/MyDrive/BreastCanser-DataSet"

CLINICAL_FILE = os.path.join(
    DATA_DIR,
    "Human__TCGA_BRCA__MS__Clinical__Clinical__01_28_2016__BI__Clinical__Firehose.tsv"
)

clinical_raw = pd.read_csv(
    CLINICAL_FILE,
    sep="\t"
)

print("="*70)
print("DATA LOADED SUCCESSFULLY")
print("="*70)

print(f"Dataset shape: {clinical_raw.shape}")
```

##Dataset Inspection

```python
# ============================================================
# Dataset Inspection
# ============================================================

print("="*70)
print("FIRST FIVE ROWS")
print("="*70)

display(clinical_raw.head())

print("\nColumn names:")
print(clinical_raw.columns.tolist()[:10])

print("\n")

print("="*70)
print("DATASET SUMMARY")
print("="*70)

print(f"Clinical variables (rows): {clinical_raw.shape[0]}")
print(f"Patients (columns):        {clinical_raw.shape[1]-1}")

print("\nFeature column:")
print(clinical_raw.columns[0])

print("\nExample patient IDs:")
print(clinical_raw.columns[1:11].tolist())

print("\nMissing values:")
print(clinical_raw.isna().sum().sum())

print("\nDuplicate clinical variables:")
print(clinical_raw.iloc[:,0].duplicated().sum())

print("\nDuplicate patient columns:")
print(clinical_raw.columns[1:].duplicated().sum())
```

##Convert Patient IDs

```python
# ============================================================
# Convert Patient IDs
# ============================================================

new_columns = [clinical_raw.columns[0]]

for col in clinical_raw.columns[1:]:
    new_columns.append(col.replace(".", "-"))

clinical_raw.columns = new_columns

print("="*70)
print("PATIENT ID CONVERSION")
print("="*70)

print("Example IDs:")

print(clinical_raw.columns[1:11].tolist())
```

##Transpose Dataset

```python
# ============================================================
# Transpose Clinical Dataset
# ============================================================

clinical_df = clinical_raw.set_index(
    clinical_raw.columns[0]
).T

clinical_df.index.name = "patient_id"

clinical_df = clinical_df.reset_index()

print("="*70)
print("PATIENT-LEVEL CLINICAL MATRIX")
print("="*70)

print(f"Patients : {clinical_df.shape[0]}")
print(f"Variables: {clinical_df.shape[1]-1}")

print("\nFirst rows:")

display(clinical_df.head())
```

##Clinical Variable Selection

```python
# ============================================================
# Clinical Variable Selection
# ============================================================

clinical = clinical_df.copy()

# Keep only medically meaningful variables
selected_columns = [
    "patient_id",
    "years_to_birth",
    "Tumor_purity",
    "pathologic_stage",
    "pathology_T_stage",
    "pathology_N_stage",
    "pathology_M_stage",
    "histological_type",
    "number_of_lymph_nodes",
    "PAM50",
    "radiation_therapy",
    "race",
    "ethnicity"
]

clinical = clinical[selected_columns]

print("="*70)
print("CLINICAL FEATURE SELECTION")
print("="*70)

print("Selected variables:")
for col in clinical.columns:
    print("-", col)

print("\nFinal shape:", clinical.shape)
```

##Missing Value Processing

```python
# ============================================================
# Missing Value Processing
# ============================================================

import numpy as np
from sklearn.impute import SimpleImputer

clinical_processed = clinical.copy()

# Replace common TCGA missing values
clinical_processed = clinical_processed.replace(
    ["NA", "na", "NaN", "NAN", "[Not Available]", "[Not Applicable]", ""],
    np.nan
)

# Numeric columns
numeric_cols = [
    "years_to_birth",
    "Tumor_purity",
    "number_of_lymph_nodes"
]

# Categorical columns
categorical_cols = [
    c for c in clinical_processed.columns
    if c not in numeric_cols + ["patient_id"]
]

# Median imputation
median_imputer = SimpleImputer(strategy="median")

clinical_processed[numeric_cols] = median_imputer.fit_transform(
    clinical_processed[numeric_cols]
)

# Most frequent imputation
mode_imputer = SimpleImputer(strategy="most_frequent")

clinical_processed[categorical_cols] = mode_imputer.fit_transform(
    clinical_processed[categorical_cols]
)

print("="*70)
print("MISSING VALUE PROCESSING")
print("="*70)

print(clinical_processed.isna().sum())
```

##Clinical Encoding

```python
# ============================================================
# Clinical Encoding
# ============================================================

from sklearn.preprocessing import OrdinalEncoder

encoded = clinical_processed.copy()

# -------------------------
# Stage Mapping
# -------------------------

stage_map = {
    "stagei":1,
    "stageia":1,
    "stageib":1,

    "stageii":2,
    "stageiia":2,
    "stageiib":2,

    "stageiii":3,
    "stageiiia":3,
    "stageiiib":3,
    "stageiiic":3,

    "stageiv":4
}

encoded["pathologic_stage"] = (
    encoded["pathologic_stage"]
    .str.lower()
    .str.replace(" ","")
    .map(stage_map)
)

# -------------------------
# T Stage
# -------------------------

t_map = {
    "tx":0,
    "t0":0,
    "tis":0,
    "t1":1,
    "t1a":1,
    "t1b":1,
    "t1c":1,
    "t2":2,
    "t3":3,
    "t4":4,
    "t4a":4,
    "t4b":4,
    "t4c":4,
    "t4d":4
}

encoded["pathology_T_stage"] = (
    encoded["pathology_T_stage"]
    .str.lower()
    .str.replace(" ","")
    .map(t_map)
)

# -------------------------
# N Stage
# -------------------------

n_map = {
    "nx":0,
    "n0":0,
    "n1":1,
    "n1a":1,
    "n1b":1,
    "n1c":1,
    "n2":2,
    "n2a":2,
    "n2b":2,
    "n3":3,
    "n3a":3,
    "n3b":3,
    "n3c":3
}

encoded["pathology_N_stage"] = (
    encoded["pathology_N_stage"]
    .str.lower()
    .str.replace(" ","")
    .map(n_map)
)

# -------------------------
# M Stage
# -------------------------

m_map = {
    "mx":0,
    "m0":0,
    "m1":1
}

encoded["pathology_M_stage"] = (
    encoded["pathology_M_stage"]
    .str.lower()
    .str.replace(" ","")
    .map(m_map)
)

# -------------------------
# Binary
# -------------------------

binary_map = {
    "yes":1,
    "no":0
}

encoded["radiation_therapy"] = (
    encoded["radiation_therapy"]
    .str.lower()
    .map(binary_map)
)

# -------------------------
# One-Hot Encoding
# -------------------------

one_hot_cols = [
    "histological_type",
    "PAM50",
    "race",
    "ethnicity"
]

encoded = pd.get_dummies(
    encoded,
    columns=one_hot_cols,
    drop_first=True
)

print("="*70)
print("ENCODING FINISHED")
print("="*70)

print("Final shape:", encoded.shape)
```

##Scaling

```python
# ============================================================
# Standardization
# ============================================================

from sklearn.preprocessing import StandardScaler

scaled = encoded.copy()

numeric_cols = [
    "years_to_birth",
    "Tumor_purity",
    "number_of_lymph_nodes"
]

scaler = StandardScaler()

scaled[numeric_cols] = scaler.fit_transform(
    scaled[numeric_cols]
)

print("="*70)
print("STANDARDIZATION")
print("="*70)

print(scaled[numeric_cols].describe())
```

##Quality Control & Save

```python
# ============================================================
# Quality Control
# ============================================================

print("="*70)
print("CLINICAL QUALITY CONTROL")
print("="*70)

print("Patients:", scaled.patient_id.nunique())

print("Features:", scaled.shape[1]-1)

print("Missing values:",
      scaled.isna().sum().sum())

print("Infinite values:",
      np.isinf(
          scaled.select_dtypes(include=np.number)
      ).sum().sum())

print("Duplicate patients:",
      scaled.patient_id.duplicated().sum())

print("\nFirst rows:")
display(scaled.head())

# ============================================================
# Save Files
# ============================================================

OUTPUT_FOLDER = "/content/drive/MyDrive/TCGA_BRCA_Project/Clinical"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

scaled.to_csv(
    f"{OUTPUT_FOLDER}/clinical_processed.csv",
    index=False
)

statistics = scaled.describe(include="all")

statistics.to_csv(
    f"{OUTPUT_FOLDER}/clinical_statistics.csv"
)

with open(
    f"{OUTPUT_FOLDER}/preprocessing_report.txt",
    "w"
) as f:

    f.write("Clinical preprocessing completed successfully.\n\n")

    f.write(f"Patients: {scaled.shape[0]}\n")
    f.write(f"Features: {scaled.shape[1]-1}\n")
    f.write(f"Missing values: {scaled.isna().sum().sum()}\n")
    f.write(f"Duplicate patients: {scaled.patient_id.duplicated().sum()}\n")

print("="*70)
print("FILES SAVED SUCCESSFULLY")
print("="*70)

print(os.listdir(OUTPUT_FOLDER))
```
