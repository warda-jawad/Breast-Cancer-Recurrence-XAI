```python
# ============================================================
# LOAD FOLLOW-UP DATASET
# ============================================================

FOLLOWUP_PATH = "/content/drive/MyDrive/BreastCanser-DataSet/follow_up.tsv"

followup_raw = pd.read_csv(
    FOLLOWUP_PATH,
    sep="\t",
    low_memory=False
)

print("="*70)
print("FOLLOW-UP DATA LOADED")
print("="*70)

print("Shape:", followup_raw.shape)
```

##Patient ID Cleaning

```python
# ============================================================
# CREATE PATIENT ID
# ============================================================

print("="*70)
print("CREATING PATIENT IDENTIFIERS")
print("="*70)


# Rename dataframe
followup_df = followup_raw.copy()


# Extract patient ID
followup_df["patient_id"] = followup_df["cases.submitter_id"]


# Convert TCGA format

followup_df["patient_id"] = (
    followup_df["patient_id"]
    .astype(str)
    .str.replace(".", "-", regex=False)
)


print("Total follow-up rows:", len(followup_df))

print(
    "Unique patients:",
    followup_df["patient_id"].nunique()
)


print("\nExample IDs:")
print(
    followup_df["patient_id"]
    .head(10)
)
```

##Keep Only Useful Columns

```python
# ============================================================
# KEEP REQUIRED FOLLOW-UP VARIABLES
# ============================================================

required_columns = [

    "patient_id",

    "follow_ups.progression_or_recurrence",

    "follow_ups.days_to_recurrence",

    "follow_ups.days_to_progression",

    "follow_ups.days_to_follow_up"

]


followup_df = followup_df[required_columns].copy()


print("="*70)
print("FOLLOW-UP VARIABLES SELECTED")
print("="*70)


print("Columns:")
print(followup_df.columns)


print("\nShape:")
print(followup_df.shape)


display(followup_df.head())
```

##Create Event Label

```python
# ============================================================
# CREATE RECURRENCE / PROGRESSION EVENT LABEL
# ============================================================

print("="*70)
print("CREATING EVENT LABEL")
print("="*70)


def clean_tcga_value(value):

    if pd.isna(value):
        return None

    value = str(value).strip().lower()

    if value in [
        "--",
        "'--",
        "",
        "nan",
        "none",
        "not reported",
        "not available"
    ]:
        return None

    return value



def valid_time(value):

    """
    Check if TCGA time value is valid
    """

    value = clean_tcga_value(value)

    if value is None:
        return False

    try:
        value = float(value)

        if value >= 0:
            return True

        return False

    except:
        return False



def create_event(row):


    progression = clean_tcga_value(
        row["follow_ups.progression_or_recurrence"]
    )


    recurrence_time = row[
        "follow_ups.days_to_recurrence"
    ]


    progression_time = row[
        "follow_ups.days_to_progression"
    ]


    # Explicit recurrence/progression

    if progression in [
        "yes",
        "true",
        "1"
    ]:
        return 1



    # Recurrence time exists

    if valid_time(recurrence_time):
        return 1



    # Progression time exists

    if valid_time(progression_time):
        return 1



    return 0



followup_df["event"] = followup_df.apply(
    create_event,
    axis=1
)



print("Event distribution:")
print(
    followup_df["event"]
    .value_counts()
)


print("\nEvent ratio:")
print(
    followup_df["event"].mean()
)
```

##Create Event Time

```python
# ============================================================
# CREATE SURVIVAL TIME (CORRECTED)
# ============================================================


print("="*70)
print("CREATING SURVIVAL TIME")
print("="*70)



def convert_time(value):


    if pd.isna(value):
        return np.nan


    value = str(value).strip()


    if value in [
        "--",
        "'--",
        "",
        "nan",
        "None"
    ]:
        return np.nan


    try:

        value = float(value)


        if value < 0:
            return np.nan


        return value


    except:

        return np.nan




# Convert columns

time_columns = [

    "follow_ups.days_to_recurrence",

    "follow_ups.days_to_progression",

    "follow_ups.days_to_follow_up"

]


for col in time_columns:

    followup_df[col] = (
        followup_df[col]
        .apply(convert_time)
    )




def extract_survival_time(row):


    # If event happened:
    # use recurrence/progression time


    if row["event"] == 1:


        if not pd.isna(
            row["follow_ups.days_to_recurrence"]
        ):
            return row[
                "follow_ups.days_to_recurrence"
            ]


        if not pd.isna(
            row["follow_ups.days_to_progression"]
        ):
            return row[
                "follow_ups.days_to_progression"
            ]



    # For censored patients:
    # use follow-up duration


    if not pd.isna(
        row["follow_ups.days_to_follow_up"]
    ):
        return row[
            "follow_ups.days_to_follow_up"
        ]


    return np.nan




followup_df["survival_time"] = followup_df.apply(
    extract_survival_time,
    axis=1
)



print("Missing survival times:")

print(
    followup_df["survival_time"]
    .isna()
    .sum()
)



print("\nSurvival statistics:")

print(
    followup_df["survival_time"]
    .describe()
)



display(
    followup_df[
        [
            "patient_id",
            "event",
            "survival_time"
        ]
    ].head()
)
```

##Convert to Patient-Level Dataset

```python
# ============================================================
# PATIENT LEVEL AGGREGATION
# ============================================================


print("="*70)
print("CREATING PATIENT LEVEL FOLLOW-UP DATASET")
print("="*70)



patient_followup = (
    followup_df
    .groupby("patient_id")
    .agg(
        {
            "event":"max",
            "survival_time":"max"
        }
    )
    .reset_index()
)



print(
    "Patients:",
    patient_followup.shape[0]
)


print(
    "\nEvent distribution:"
)


print(
    patient_followup["event"]
    .value_counts()
)


display(
    patient_followup.head()
)
```

##Quality Control

```python
# ============================================================
# FINAL SURVIVAL CLEANING
# ============================================================

print("="*70)
print("FINAL SURVIVAL CLEANING")
print("="*70)


print("Original shape:")
print(patient_followup.shape)


# ============================================================
# Remove patients without survival time
# ============================================================

followup_final = patient_followup[
    patient_followup["survival_time"].notna()
].copy()


print("\nAfter removing missing survival:")
print(followup_final.shape)



# ============================================================
# Fix zero or negative survival times
# ============================================================

followup_final.loc[
    followup_final["survival_time"] <= 0,
    "survival_time"
] = 1



# ============================================================
# FINAL QUALITY CONTROL
# ============================================================

print("\n")
print("="*70)
print("FINAL FOLLOW-UP QUALITY CONTROL")
print("="*70)


print(
    "Patients:",
    followup_final["patient_id"].nunique()
)


print(
    "Duplicate patients:",
    followup_final["patient_id"].duplicated().sum()
)


print(
    "Missing event:",
    followup_final["event"].isna().sum()
)


print(
    "Missing survival time:",
    followup_final["survival_time"].isna().sum()
)



print("\nEvent distribution:")
print(
    followup_final["event"].value_counts()
)


print("\nEvent ratio:")
print(
    followup_final["event"].mean()
)


display(
    followup_final.head()
)
```

##Save Final Files

```python
# ============================================================
# SAVE FINAL FOLLOW-UP DATASETS
# ============================================================


import os


SAVE_DIR = "/content/drive/MyDrive/TCGA_BRCA/followup_processed"


os.makedirs(
    SAVE_DIR,
    exist_ok=True
)



# ============================================================
# Save final patient-level follow-up dataset
# ============================================================

followup_final.to_csv(
    f"{SAVE_DIR}/followup_processed.csv",
    index=False
)



# ============================================================
# Save event labels
# ============================================================

followup_final[
    [
        "patient_id",
        "event"
    ]
].to_csv(
    f"{SAVE_DIR}/followup_event_labels.csv",
    index=False
)



# ============================================================
# Save final statistics
# ============================================================

followup_final[
    [
        "event",
        "survival_time"
    ]
].describe().to_csv(
    f"{SAVE_DIR}/followup_statistics.csv"
)



print("="*70)
print("FINAL FOLLOW-UP FILES SAVED")
print("="*70)


print(
    os.listdir(SAVE_DIR)
)



print("\nFinal dataset shape:")
print(
    followup_final.shape
)
```
