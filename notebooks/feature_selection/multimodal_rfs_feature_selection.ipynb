# Feature Selection Pipeline

## Overview

This notebook performs feature selection for a multimodal breast cancer Recurrence-Free Survival (RFS) prediction dataset.

The objective is to reduce genomic dimensionality while preserving clinically relevant information and preparing the dataset for survival analysis models.

Feature selection is performed using only the training dataset and subsequently applied to validation and test datasets to prevent data leakage.

---

## Pipeline

1. Load training, validation, and test datasets.

2. Separate clinical and genomic modalities.

3. Remove constant genomic features using Variance Threshold.

4. Remove highly correlated genomic features using Pearson correlation filtering (threshold > 0.95).

5. Select the top 200 genomic features using univariate Cox proportional hazards regression based on RFS survival outcome.

6. Apply the selected genomic feature set to validation and test datasets.

7. Save final selected datasets and selected gene list.

```python
# ============================================================
# FEATURE GROUP IDENTIFICATION
# ============================================================


print("="*70)
print("IDENTIFYING FEATURE GROUPS")
print("="*70)



# Fixed columns

non_feature_columns = [
    "patient_id",
    "file_uuid",
    "event",
    "survival_time"
]



# Clinical features
clinical_features = [
    "years_to_birth",
    "Tumor_purity",
    "pathologic_stage",
    "pathology_T_stage",
    "pathology_N_stage",
    "pathology_M_stage",
    "number_of_lymph_nodes",
    "radiation_therapy"
]


clinical_features += [
    c for c in train.columns
    if c.startswith(
        (
            "histological_type",
            "PAM50",
            "race",
            "ethnicity"
        )
    )
]



# Genomic features

exclude_columns = (
    non_feature_columns
    +
    clinical_features


)


genomic_features = [
    c for c in train.columns
    if c not in exclude_columns
]



print(
    "Clinical features:",
    len(clinical_features)
)


print(
    "Genomic features:",
    len(genomic_features)
)


print(
    "First genomic features:"
)


print(
    genomic_features[:20]
)
```

##Remove Constant Genes

```python
# ============================================================
# REMOVE CONSTANT GENOMIC FEATURES
# ============================================================


X_train_genes = train[genomic_features]


variance_selector = VarianceThreshold(
    threshold=0
)


X_train_variance = variance_selector.fit_transform(
    X_train_genes
)



selected_after_variance = (
    X_train_genes.columns[
        variance_selector.get_support()
    ]
)



print("="*70)
print("VARIANCE FILTER")
print("="*70)


print(
    "Original genes:",
    len(genomic_features)
)


print(
    "Remaining genes:",
    len(selected_after_variance)
)
```

##Remove Highly Correlated Genes

```python
# ============================================================
# REMOVE HIGHLY CORRELATED GENES
# ============================================================


corr_matrix = (
    train[selected_after_variance]
    .corr()
)



upper_triangle = (
    corr_matrix
    .where(
        np.triu(
            np.ones(
                corr_matrix.shape
            ),
            k=1
        ).astype(bool)
    )
)



correlated_features = [
    column
    for column in upper_triangle.columns
    if any(
        upper_triangle[column] > 0.95
    )
]



genes_after_corr = [
    g for g in selected_after_variance
    if g not in correlated_features
]



print("="*70)
print("CORRELATION FILTER")
print("="*70)


print(
    "Removed:",
    len(correlated_features)
)


print(
    "Remaining:",
    len(genes_after_corr)
)
```

##COX SURVIVAL Feature Selection

```python
# ============================================================
# COX SURVIVAL FEATURE SELECTION
# ============================================================

from lifelines import CoxPHFitter


TOP_K = 200


cox_results = []


for gene in genes_after_corr:

    temp = train[
        [
            gene,
            "survival_time",
            "event"
        ]
    ].copy()


    try:

        cph = CoxPHFitter()

        cph.fit(
            temp,
            duration_col="survival_time",
            event_col="event"
        )


        p_value = (
            cph.summary
            .loc[gene, "p"]
        )


        cox_results.append(
            {
                "gene": gene,
                "p_value": p_value
            }
        )


    except:

        continue



cox_results = pd.DataFrame(
    cox_results
)



cox_results = (
    cox_results
    .sort_values(
        "p_value"
    )
)



selected_genes = (
    cox_results
    .head(TOP_K)["gene"]
    .tolist()
)


print("="*70)
print("COX FEATURE SELECTION")
print("="*70)


print(
    "Genes tested:",
    len(cox_results)
)


print(
    "Selected genes:",
    len(selected_genes)
)
```

##Apply Selection to All Splits

```python
# ============================================================
# APPLY FEATURE SELECTION
# ============================================================


common_columns = (
    clinical_features
    +
    selected_genes
    +
    [
        "patient_id",
        "file_uuid",
        "event",
        "survival_time"
    ]
)



train_selected = train[
    common_columns
]


validation_selected = validation[
    common_columns
]


test_selected = test[
    common_columns
]



print(train_selected.shape)

print(validation_selected.shape)

print(test_selected.shape)
```
